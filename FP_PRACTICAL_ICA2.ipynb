```python

#Question 1 set 6 siona
sentence = input("Enter a sentence: ")
with_spaces = len(sentence)6
without_space = len(sentence.replace("",""))
words = sentence.split()
num_words = len(words)
upper_case = sentence.upper()
lower_case = sentence.lower()
first_word = words[0]
last_word = words[-1]
title_case = sentence.title()

#display the result
print("characters with spaces:", with_spaces)
print("charcters without spaces:", without_space)
print("Number of words: ",num_words)
print("Uppercase: ", upper_case)
print("Lowercase : ", lower_case)
print("Firstword : ", first_word)
print("Lastword : ", last_word)
print("titlecase :", title_case)

#question 2 set 6 siona


# 1. Store the names of 5 students and their percentages in two lists
student_names = ["Aarav", "Diya", "Kabir", "Meera", "Rohan"]
student_percentages = [94.5, 82.0, 68.5, 45.0, 38.5]


# 2. Function to return the grade based on the percentage
def percentage(score):
  if score >= 90:
    return "A+"
  elif score >= 75:
    return "A"
  elif score >= 60:
    return "b"
  elif score >= 40:
    return "c"
  else:
    return "f"


# Variables to track the topper and failure count
topper_name = ""
highest_score = -1
fail_count = 0

# 3. Print the formatted result table header
print("-" * 45)
print(f"{'Student Name':<18} | {'Percentage':<12} | {'Grade':<6}")
print("-" * 45)

# 4. Use a for loop to process each student
for i in range(len(student_names)):
  name = student_names[i]
  score = student_percentages[i]
  grade = percentage(score)  # Call the function

  # Print formatted row
  print(f"{name:<18} | {score:<12.2f} | {grade:<6}")

  # Check for topper
  if score > highest_score:
    highest_score = score
    topper_name = name

  # Check for failures (below 40)
  if score < 40:
    fail_count += 1

print("-" * 45)

# 5. Display summary results
print(f"\n🏆 Topper: **{topper_name}** with **{highest_score}%**")
print(f"⚠️ Total number of students who failed: **{fail_count}**")







```
